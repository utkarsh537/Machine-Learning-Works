```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/wine_data.csv', header=None, usecols=[0,1,2])
df.columns = ['Class label', 'Alcohol', 'Malic acid']
```

```python
df.head()
```

```python
sns.kdeplot(df['Alcohol'], shade=True)
```

```python
sns.kdeplot(df['Malic acid'], shade= True)
```

```python
color_dict = {1: 'red', 2: 'blue', 3: 'green'}
sns.scatterplot(x='Alcohol', y='Malic acid', data=df, hue='Class label',palette=color_dict )
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('Class label' ,axis =1), df['Class label'], test_size=0.3, random_state=0)
X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit (X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
np.round(X_train_scaled.describe(), 1)
```

```python
fig,(ax1,ax2) = plt.subplots(ncols = 2, figsize=(12,5))
ax1.scatter(X_train['Alcohol'],X_train['Malic acid'], c = y_train, )
ax1.set_title('Original Data')
ax2.scatter(X_train_scaled['Alcohol'],X_train_scaled['Malic acid'], c = y_train)
ax2.set_title('Scaled Data')
plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(ncols=2, figsize=(12,5))
ax1.set_title ('Original Data')
sns.kdeplot(X_train['Alcohol'], shade=True, ax=ax1)
sns.kdeplot(X_train['Malic acid'], shade=True, ax=ax1)
ax2.set_title ('Scaled Data')
sns.kdeplot(X_train_scaled['Alcohol'], shade=True, ax=ax2)
sns.kdeplot(X_train_scaled['Malic acid'], shade=True, ax=ax2)
plt.show()
```
